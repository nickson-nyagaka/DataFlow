**CHECK FOR NON MATCHING CCC NUMBERS**

```python
# Robust CCC/NUPI mismatch check (memory-aware, preserves formats, drops null left IDs)
import numpy as np
import pandas as pd

left_key = 'ccc_number_nupi_n_a_10_digit_number_'
right_key = 'ccc_number_nupi_n_a_10_digit_number_'

# small helper: normalize but preserve index (remove trailing .0 from floats and trim)
def _normalize_ids(ser: pd.Series) -> pd.Series:
    s = ser.astype(str).str.strip()
    s = s.replace({'': pd.NA, 'nan': pd.NA})
    s = s.dropna()
    # common Excel float artifact: '12345.0' -> '12345'
    s = s.str.replace(r"\.0+$", "", regex=True)
    return s

if left_key not in ConsolidatedKPLHIV.columns or right_key not in ConsolidatedFSW.columns:
    missing = []
    if left_key not in ConsolidatedKPLHIV.columns:
        missing.append(left_key)
    if right_key not in ConsolidatedFSW.columns:
        missing.append(right_key)
    print(f"Missing key column(s): {', '.join(missing)}")
else:
    # normalize right-side IDs (drop blanks/nulls)
    right_ser = _normalize_ids(ConsolidatedFSW[right_key])

    # threshold to decide whether to build a set or use a merge-based approach
    SET_THRESHOLD = 500_000

    if right_ser.nunique() <= SET_THRESHOLD:
        # fast set membership (memory okay for moderate unique counts)
        right_ids = set(right_ser.unique())

        # normalize left-side but keep index so we can map results back
        left_norm = _normalize_ids(ConsolidatedKPLHIV[left_key])

        # mask only for non-null left IDs; do NOT count missing left IDs as non-matching
        mask_non_matching = pd.Series(False, index=ConsolidatedKPLHIV.index)
        if not left_norm.empty:
            mask_non_matching[left_norm.index] = ~left_norm.isin(right_ids)
    else:
        # right side is very large — use a merge with indicator which is more memory-friendly
        right_df = right_ser.drop_duplicates().to_frame(name='_norm')

        left_df = ConsolidatedKPLHIV[[left_key]].copy()
        left_df['_norm'] = _normalize_ids(left_df[left_key]).reindex(left_df.index)

        # reset index to keep original index after merge
        left_df = left_df.reset_index()
        merged = left_df.merge(right_df.reset_index(drop=True), on='_norm', how='left', indicator=True)

        mask_non_matching = pd.Series(False, index=ConsolidatedKPLHIV.index)
        if '_merge' in merged.columns:
            left_only = merged[merged['_merge'] == 'left_only']
            mask_non_matching[left_only['index']] = True

    non_matching_cc = ConsolidatedKPLHIV[mask_non_matching].copy()
    print(f"Found {len(non_matching_cc):,} records in ConsolidatedKPLHIV not in ConsolidatedFSW (by {left_key}).")

    # optional: inspect a few columns to validate without showing NaNs
    display_cols = [c for c in [left_key, 'kp_vp_unique_identifier_code_', 'region', 'sr_name'] if c in non_matching_cc.columns]
    if display_cols:
        display_df = non_matching_cc[display_cols].fillna('')
        try:
            display(display_df.head(1000))
        except Exception:
            print(display_df.head(1000))
```

```python
# Standardize function
def standardize_viral_load(val):
    val = str(val).strip().lower()

    # Normalize anything like LDL, >LDL, <LDL
    if re.fullmatch(r'[<>]?ldl', val):
        return "LDL"

    # Extract numeric part (e.g., from "9250cp/ml", "211000 copies")
    match = re.search(r'[\d,]+', val)
    if match:
        return int(match.group().replace(',', ''))

    return "UNKNOWN"

# Apply transformation
ConsolidatedKPLHIV["standardized_viral_load"] = ConsolidatedKPLHIV["levels_copies_ml_"].apply(standardize_viral_load)
def is_suppressed(val):
    if val == "LDL":
        return True
    elif isinstance(val, int) and val < 200:
        return True
    return False
ConsolidatedKPLHIV['is_suppressed'] = ConsolidatedKPLHIV['standardized_viral_load'].apply(is_suppressed)
summary_KPLHIV = ConsolidatedKPLHIV.groupby(['county', 'sr_name', 'region']).agg(
    Total_KPLHIV=('kp_vp_unique_identifier_code_', 'nunique'),
    Suppressed=('is_suppressed','sum'),
).reset_index()

summary_KPLHIV
```

## **CALCULATE COVERAGE INDICATORS**

```python
import re

# Helper: get columns by regex pattern
def get_cols(df, pattern):
    return [col for col in df.columns if re.fullmatch(pattern, col)]

# Helper: any value in values for columns (mark all rows for a client if any row matches)
def any_in_unique_allrows(df, cols, values, id_col="kp_vp_unique_identifier_code"):
    if not cols:
        return pd.Series(False, index=df.index)
    row_has = df[cols].isin(values).any(axis=1)
    # For each unique_id, check if ANY row had the event, mark all rows for that id
    has_by_id = row_has.groupby(df[id_col]).transform("max")
    return has_by_id

# =====================
# 1. PrEP indicators
# =====================
prep_cols = get_cols(ConsolidatedFSW, r"prep_status_?\d*$")
prep_cols = sorted(prep_cols)
prep_q1_cols = prep_cols[:3]
prep_q2_cols = prep_cols[3:6]

ConsolidatedFSW['prep_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q1_cols, ['Initiated','Restart','Switched','Refill']), 1, 0)
ConsolidatedFSW['prep_refill_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q1_cols, ['Refill']), 1, 0)
ConsolidatedFSW['prep_initiate_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q1_cols, ['Initiated']), 1, 0)
ConsolidatedFSW['prep_disc_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q1_cols, ['Discontinued']), 1, 0)

ConsolidatedFSW['prep_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q2_cols, ['Initiated', 'Restart','Switched','Refill']), 1, 0)
ConsolidatedFSW['prep_refill_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q2_cols, ['Refill']), 1, 0)
ConsolidatedFSW['prep_initiate_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q2_cols, ['Initiated']), 1, 0)
ConsolidatedFSW['prep_disc_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, prep_q2_cols, ['Discontinued']), 1, 0)
ConsolidatedFSW['prep_sem'] = np.where((ConsolidatedFSW['prep_q1'] + ConsolidatedFSW['prep_q2']) >= 1, 1, 0)

# =====================
# 2. HIV testing indicators
# =====================
hiv_cols = get_cols(ConsolidatedFSW, r"hiv_tested_results_p_n_nd_na_?\d*$")
hiv_cols = sorted(hiv_cols)
hiv_q1_cols = hiv_cols[:3]
hiv_q2_cols = hiv_cols[3:6]

ConsolidatedFSW['hts_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q1_cols, ['Negative', 'Positive']), 1, 0)
ConsolidatedFSW['hts_Negative_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q1_cols, ['Negative']), 1, 0)
ConsolidatedFSW['hts_Positive_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q1_cols, ['Positive']), 1, 0)

ConsolidatedFSW['hts_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q2_cols, ['Negative', 'Positive']), 1, 0)
ConsolidatedFSW['hts_Negative_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q2_cols, ['Negative']), 1, 0)
ConsolidatedFSW['hts_Positive_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q2_cols, ['Positive']), 1, 0)

# ensure date column is datetime
ConsolidatedFSW['date_of_enrolment'] = pd.to_datetime(
    ConsolidatedFSW['date_of_enrolment'], errors='coerce'
)

# boolean: enrolled in Apr or later (month >= 10)
enrolled_Apr_plus = ConsolidatedFSW['date_of_enrolment'].dt.month >= 7

# new HTS_SEM: (both Q1 & Q2) OR (Q2 and enrolled Apr+)
ConsolidatedFSW['hts_sem'] = np.where(
    (
        (ConsolidatedFSW['hts_q1'] > 0) & (ConsolidatedFSW['hts_q2'] > 0)
    ) |
    (
        (ConsolidatedFSW['hts_q2'] > 0) & (enrolled_Apr_plus)
    ),
    1,
    0
)
ConsolidatedFSW['hts_sem_any'] = np.where(
    ((ConsolidatedFSW['hts_q1'] > 0) | (ConsolidatedFSW['hts_q2'] > 0)) ,
    1, 0
)

# =====================
# 3. ART outcome indicators
# =====================
art_cols = get_cols(ConsolidatedFSW, r"art_outcome_?\d*$")
art_cols = sorted(art_cols)
art_q1_cols = art_cols[:3]
art_q2_cols = art_cols[3:6]

ConsolidatedFSW['activeart'] = np.where(any_in_unique_allrows(ConsolidatedFSW, art_q1_cols, ['Active', 'Yes']), 1, 0)
ConsolidatedFSW['activeart_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, art_q2_cols, ['Active', 'Yes']), 1, 0)

# =====================
# 4. STI indicators
# =====================
sti_screened_cols = get_cols(ConsolidatedFSW, r"screened_for_stis_y_n_?\d*$")
sti_screened_cols = sorted(sti_screened_cols)
sti_diag_cols = get_cols(ConsolidatedFSW, r"diagnosed_with_sti_y_n_?\d*$")
sti_diag_cols = sorted(sti_diag_cols)
sti_treat_cols = get_cols(ConsolidatedFSW, r"treated_for_stis_y_n_na_r_referred_?\d*$")
sti_treat_cols = sorted(sti_treat_cols)

sti_screened_q1_cols = sti_screened_cols[:3]
sti_screened_q2_cols = sti_screened_cols[3:6]
sti_diag_q1_cols = sti_diag_cols[:3]
sti_diag_q2_cols = sti_diag_cols[3:6]
sti_treat_q1_cols = sti_treat_cols[:3]
sti_treat_q2_cols = sti_treat_cols[3:6]

ConsolidatedFSW['sti_screened_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, sti_screened_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['sti_screened_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, sti_screened_q2_cols, ['Yes']), 1, 0)
ConsolidatedFSW['sti_screened_sem'] = np.where((ConsolidatedFSW['sti_screened_q1'] + ConsolidatedFSW['sti_screened_q2']) >= 1, 1, 0)

ConsolidatedFSW['sti_diagnosed_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, sti_diag_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['sti_diagnosed_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, sti_diag_q2_cols, ['Yes']), 1, 0)
ConsolidatedFSW['sti_diagnosed_sem'] = np.where((ConsolidatedFSW['sti_diagnosed_q1'] + ConsolidatedFSW['sti_diagnosed_q2']) >= 1, 1, 0)

ConsolidatedFSW['sti_treated_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, sti_treat_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['sti_treated_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, sti_treat_q2_cols, ['Yes']), 1, 0)
ConsolidatedFSW['sti_treated_sem'] = np.where((ConsolidatedFSW['sti_treated_q1'] + ConsolidatedFSW['sti_treated_q2']) >= 1, 1, 0)

# =====================
# 5. Peer Education & RRC
# =====================
peer_ed_cols = get_cols(ConsolidatedFSW, r"received_peer_education_y_n_?\d*$")
peer_ed_cols = sorted(peer_ed_cols)
rrc_cols = get_cols(ConsolidatedFSW, r"provided_with_risk_reduction_counselling_y_n_?\d*$")
rrc_cols = sorted(rrc_cols)

peer_ed_q1_cols = peer_ed_cols[:3]
peer_ed_q2_cols = peer_ed_cols[3:6]
rrc_q1_cols = rrc_cols[:3]
rrc_q2_cols = rrc_cols[3:6]

ConsolidatedFSW['peereducation_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, peer_ed_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['peereducation_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, peer_ed_q2_cols, ['Yes']), 1, 0)
ConsolidatedFSW['peereducation_sem'] = np.where((ConsolidatedFSW['peereducation_q1'] + ConsolidatedFSW['peereducation_q2']) >= 1, 1, 0)

ConsolidatedFSW['rrc_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, rrc_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['rrc_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, rrc_q2_cols, ['Yes']), 1, 0)

ConsolidatedFSW['he_rrc_q1'] = np.where((ConsolidatedFSW['peereducation_q1'] == 1) | (ConsolidatedFSW['rrc_q1'] == 1), 1, 0)
ConsolidatedFSW['he_rrc_q2'] = np.where((ConsolidatedFSW['peereducation_q2'] == 1) | (ConsolidatedFSW['rrc_q2'] == 1), 1, 0)

# =====================
# 6. TB screening indicators
# =====================
tb_screened_cols = get_cols(ConsolidatedFSW, r"screened_for_tb_?\d*$")
tb_screened_cols = sorted(tb_screened_cols)

tb_screened_q1_cols = tb_screened_cols[:3]
tb_screened_q2_cols = tb_screened_cols[3:6]

ConsolidatedFSW['tb_screened_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, tb_screened_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['tb_screened_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, tb_screened_q2_cols, ['Yes']), 1, 0)
ConsolidatedFSW['tb_screened_sem'] = np.where((ConsolidatedFSW['tb_screened_q1'] + ConsolidatedFSW['tb_screened_q2']) >= 1, 1, 0)

# =====================
# 7. Condoms & NSP
# =====================
condoms_cols = get_cols(ConsolidatedFSW, r"condoms_distributed_indicate_number_n_a_for_inmates_?\d*$")
condoms_cols = sorted(condoms_cols)
nsp_cols = get_cols(ConsolidatedFSW, r"needles_and_syringes_distributed_indicate_number_n_a_for_inmates_?\d*$")
nsp_cols = sorted(nsp_cols)

condoms_q1_cols, condoms_q2_cols = condoms_cols[:3], condoms_cols[3:6]
nsp_q1_cols, nsp_q2_cols = nsp_cols[:3], nsp_cols[3:6]

# Convert to numeric safely
if condoms_cols or nsp_cols:
    ConsolidatedFSW[condoms_cols + nsp_cols] = ConsolidatedFSW[condoms_cols + nsp_cols].apply(pd.to_numeric, errors='coerce')

ConsolidatedFSW['condoms_q1'] = np.where(ConsolidatedFSW[condoms_q1_cols].gt(0).any(axis=1) if condoms_q1_cols else False, 1, 0)
ConsolidatedFSW['condoms_q2'] = np.where(ConsolidatedFSW[condoms_q2_cols].gt(0).any(axis=1) if condoms_q2_cols else False, 1, 0)
ConsolidatedFSW['condoms_sem'] = np.where((ConsolidatedFSW['condoms_q1'] + ConsolidatedFSW['condoms_q2']) >= 1, 1, 0)

ConsolidatedFSW['total_condoms_q1'] = ConsolidatedFSW[condoms_q1_cols].sum(axis=1, skipna=True) if condoms_q1_cols else 0
ConsolidatedFSW['total_condoms_q2'] = ConsolidatedFSW[condoms_q2_cols].sum(axis=1, skipna=True) if condoms_q2_cols else 0
ConsolidatedFSW['total_condoms_sem'] = ConsolidatedFSW['total_condoms_q1'] + ConsolidatedFSW['total_condoms_q2']
ConsolidatedFSW['nsp_q1'] = np.where(ConsolidatedFSW[nsp_q1_cols].gt(0).any(axis=1) if nsp_q1_cols else False, 1, 0)
ConsolidatedFSW['nsp_q2'] = np.where(ConsolidatedFSW[nsp_q2_cols].gt(0).any(axis=1) if nsp_q2_cols else False, 1, 0)
ConsolidatedFSW['nsp_sem'] = np.where((ConsolidatedFSW['nsp_q1'] + ConsolidatedFSW['nsp_q2']) >= 1, 1, 0)

ConsolidatedFSW['total_nsp_q1'] = ConsolidatedFSW[nsp_q1_cols].sum(axis=1, skipna=True) if nsp_q1_cols else 0
ConsolidatedFSW['total_nsp_q2'] = ConsolidatedFSW[nsp_q2_cols].sum(axis=1, skipna=True) if nsp_q2_cols else 0

# =====================
# 8. Positive / ART / PLHIV
# =====================

onart_cols = get_cols(ConsolidatedFSW, r"currently_on_art_y_n_n_a_?\d*$")
onart_cols = sorted(onart_cols)
onart_q1_cols = onart_cols[:3]
onart_q2_cols = onart_cols[3:6]

ConsolidatedFSW['onart_q1'] = np.where(any_in_unique_allrows(ConsolidatedFSW, onart_q1_cols, ['Yes']), 1, 0)
ConsolidatedFSW['onart_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, onart_q2_cols, ['Yes']), 1, 0)
ConsolidatedFSW['onart_sem'] = np.where((ConsolidatedFSW['onart_q1'] + ConsolidatedFSW['onart_q2']) >= 1, 1, 0)

ConsolidatedFSW['Positive_art'] = np.where((ConsolidatedFSW['hts_Positive_q1'] == 1) & (ConsolidatedFSW['onart_q1'] == 1), 1, 0)
ConsolidatedFSW['Positive_art_q2'] = np.where((ConsolidatedFSW['hts_Positive_q2'] == 1) & (ConsolidatedFSW['onart_q2'] == 1), 1, 0)

ConsolidatedFSW['plhiv'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q1_cols, ['Known positive']), 1, 0)
ConsolidatedFSW['plhiv_q2'] = np.where(any_in_unique_allrows(ConsolidatedFSW, hiv_q2_cols, ['Known positive']), 1, 0)

# =====================
# 9. PLHIV Service
# =====================
ConsolidatedFSW['plhivservice_q1'] = np.where(
    (ConsolidatedFSW['plhiv'] == 1) &
    (ConsolidatedFSW['sti_screened_q1'] == 1) &
    (ConsolidatedFSW['condoms_q1'] == 1) &
    # (ConsolidatedFSW['nsp_q1'] == 1) &
    (ConsolidatedFSW['he_rrc_q1'] == 1) &
    (ConsolidatedFSW['tb_screened_q1'] == 1),
    1, 0
)

ConsolidatedFSW['plhivservice_q2'] = np.where(
    (ConsolidatedFSW['plhiv_q2'] == 1) &
    (ConsolidatedFSW['sti_screened_q2'] == 1) &
    (ConsolidatedFSW['condoms_q2'] == 1) &
    # (ConsolidatedFSW['nsp_q2'] == 1) &
    (ConsolidatedFSW['he_rrc_q2'] == 1) &
    (ConsolidatedFSW['tb_screened_q2'] == 1),
    1, 0
)

# =====================
# 10. Defined Package
# =====================
ConsolidatedFSW['definedpackage_q1'] = np.where(
    ((ConsolidatedFSW['hts_q1'] == 1) &
     (ConsolidatedFSW['sti_screened_q1'] == 1) &
     (ConsolidatedFSW['tb_screened_q1'] == 1) &
    #  (ConsolidatedFSW['nsp_q1'] == 1) &
     (ConsolidatedFSW['condoms_q1'] == 1) &
     (ConsolidatedFSW['he_rrc_q1'] == 1)) |
    (ConsolidatedFSW['plhivservice_q1'] == 1),
    1, 0
)

ConsolidatedFSW['definedpackage_q2'] = np.where(
    ((ConsolidatedFSW['hts_q2'] == 1) &
     (ConsolidatedFSW['sti_screened_q2'] == 1) &
     (ConsolidatedFSW['tb_screened_q2'] == 1) &
    #  (ConsolidatedFSW['nsp_q2'] == 1) &
     (ConsolidatedFSW['condoms_q2'] == 1) &
     (ConsolidatedFSW['he_rrc_q2'] == 1)) |
    (ConsolidatedFSW['plhivservice_q2'] == 1),
    1, 0
)

ConsolidatedFSW['definedpackage_sem'] = np.where((ConsolidatedFSW['definedpackage_q1'] + ConsolidatedFSW['definedpackage_q2']) >= 1, 1, 0)
```

```python
# Group by and count occurrences for each age group, county, sr_name, region
age_group_counts = (
    ConsolidatedFSW
    .groupby(['county', 'sr_name', 'region', 'age_group'])
    .size()
    .reset_index(name='total')
)

# Pivot the table to make age_group columns
age_group_counts = (
    age_group_counts
    .pivot_table(
        index=['county', 'sr_name', 'region'],
        columns='age_group',
        values='total',
        fill_value=0
    )
    .reset_index()
)

# Summarizing the main data by county, sr_name, region
summary_FSW = (
    ConsolidatedFSW
    .groupby(['county', 'sr_name', 'region'])
    .agg(
        TotalCohort=('kp_vp_unique_identifier_code', 'nunique'),
        TotalPE=('peer_educator_code', 'nunique'),
        HTSQ1=('hts_q1', 'sum'),
        HTSQ2=('hts_q2', 'sum'),
        HTS_SEM=('hts_sem', 'sum'),
        HTS_SEM_ANY=('hts_sem_any', 'sum'),
        HTS_Negative=('hts_Negative_q1', 'sum'),
        HTS_Negative_Q2=('hts_Negative_q2', 'sum'),
        HTS_Positive=('hts_Positive_q1', 'sum'),
        HTS_Positive_Q2=('hts_Positive_q2', 'sum'),

        DefinedPackage_Q1=('definedpackage_q1', 'sum'),
        DefinedPackage_Q2=('definedpackage_q2', 'sum'),
        DefinedPackage_SEM=('definedpackage_sem', 'sum'),

        New_on_ART=('Positive_art', 'sum'),
        New_on_ART_Q2=('Positive_art_q2', 'sum'),

        PLHIV_Q1=('plhiv', 'sum'),
        PLHIV_Q2=('plhiv_q2', 'sum'),

        CurrentArt_Q1=('onart_q1', 'sum'),
        CurrentArt_Q2=('onart_q2', 'sum'),

        ReceivedCondom_Q1=('condoms_q1', 'sum'),
        ReceivedCondom_Q2=('condoms_q2', 'sum'),
        ReceivedCondom_SEM=('condoms_sem', 'sum'),

        STISCREENEDQ_1=('sti_screened_q1', 'sum'),
        STISCREENEDQ_2=('sti_screened_q2', 'sum'),
        STI_SCREENED_SEM=('sti_screened_sem', 'sum'),

        STIDiagnosid_Q1=('sti_diagnosed_q1', 'sum'),
        STI_DIAGNOSED_Q2=('sti_diagnosed_q2', 'sum'),
        STI_DIAGNOSED_SEM=('sti_diagnosed_sem', 'sum'),

        STItreated_Q1=('sti_treated_q1', 'sum'),
        STI_TREATED_Q2=('sti_treated_q2', 'sum'),
        STI_TREATED_SEM=('sti_treated_sem', 'sum'),

        TB_SCREENED_Q1=('tb_screened_q1', 'sum'),
        TB_SCREENED_Q2=('tb_screened_q2', 'sum'),
        TB_SCREENED_SEM=('tb_screened_sem', 'sum'),

        PeerEducation_Q1=('peereducation_q1', 'sum'),
        PeerEducation_Q2=('peereducation_q2', 'sum'),
        PeerEducation_SEM=('peereducation_sem', 'sum'),

        HE_RRC_Q1=('he_rrc_q1', 'sum'),
        HE_RRC_Q2=('he_rrc_q2', 'sum'),

        PREP_Q1=('prep_q1', 'sum'),
        PREP_Q2=('prep_q2', 'sum'),
        PREP_SEM=('prep_sem', 'sum'),

        NSP_Q1=('nsp_q1', 'sum'),
        NSP_Q2=('nsp_q2', 'sum'),
        NSP_SEM=('nsp_sem', 'sum'),

        total_condoms_Q1=('total_condoms_q1', 'sum'),
        total_condoms_Q2=('total_condoms_q2', 'sum'),
        total_condoms_SEM=('total_condoms_sem', 'sum'),
        total_nsp_Q1=('total_nsp_q1', 'sum'),
        total_nsp_Q2=('total_nsp_q2', 'sum')
    )
    .reset_index()
)

# --- Merge age group counts with summary_FSW based on common columns ---
final_summary_FSW = pd.merge(
    summary_FSW,
    age_group_counts,
    on=['county', 'sr_name', 'region'],
    how='left'
)

keys = ['county', 'sr_name', 'region']
# Age group cols = all columns that came from age_group_counts except keys
age_group_cols = [
    c for c in age_group_counts.columns
    if c not in keys and c in final_summary_FSW.columns
]
cols = list(final_summary_FSW.columns)
# Remaining columns: everything that is not key, not age group, not cohort
remaining = [
    c for c in cols
    if c not in keys + age_group_cols
]
# New order: keys -> age groups -> cohort cols -> remaining
new_order = keys + age_group_cols  + remaining

# Apply final ordering (filter to existing columns, avoid duplicates)
seen = set()
ordered = []
for c in new_order:
    if c in final_summary_FSW.columns and c not in seen:
        seen.add(c)
        ordered.append(c)

final_summary_FSW = final_summary_FSW[ordered]

final_summary_FSW
```

```python
import os
import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import Alignment, Font, PatternFill, Border, Side
from openpyxl.utils import get_column_letter

# -------------------------
# Color constants — resolved from the actual FSW template (theme colors + tint decoded)
# -------------------------
SUBHEADER_FILL_COLOR = "FF7D81"   # coral — row 2, uniform across every block
TOTAL_FILL_COLOR = "6BD669"       # green — TOTAL row

GROUP_HEADER_COLORS = {
    "COUNTY": "00B050",
    "SR NAME": "00B050",
    "REGION": "00B050",
    "AGE GROUP": "B7DEE8",
    "TOTAL": "00B050",
    "HIV TESTING & PLHIV": "F79646",
    "DEFINED PACKAGE": "0070C0",
    "PREP ": "FCD5B5",
    "ART": "8064A2",
    "STI SCREENING": "EBF1DE",
    "STI DIAGNOSIS": "4BACC6",
    "STI TREATMENT": "FFC000",
    "TB SCREENING": "B3A2C7",
    "PEER EDUCATION": "FFFF00",
    "CONDOMS": "D9D9D9",
}
DEFAULT_GROUP_COLOR = "FF7D81"  # fallback for any block name not in the map above

SUBHEADER_FILL = PatternFill("solid", fgColor=SUBHEADER_FILL_COLOR)
TOTAL_FILL = PatternFill("solid", fgColor=TOTAL_FILL_COLOR)

# -------------------------
# Font constants
# -------------------------
HEADER_FONT_NAME = "Cascadia Code"
BODY_FONT_NAME = "Candara"

HEADER_FONT = Font(name=HEADER_FONT_NAME, bold=True)
BODY_FONT = Font(name=BODY_FONT_NAME)
BODY_FONT_BOLD = Font(name=BODY_FONT_NAME, bold=True)


def build_age_group_counts(df: pd.DataFrame) -> pd.DataFrame:

    age_group_counts = (
        df.groupby(["county", "sr_name", "region", "age_group"])
          .size()
          .reset_index(name="total")
          .pivot_table(
              index=["county", "sr_name", "region"],
              columns="age_group",
              values="total",
              fill_value=0,
              aggfunc="sum"
          )
          .reset_index()
    )
    return age_group_counts


def build_summary_fsw(ConsolidatedFSW: pd.DataFrame) -> pd.DataFrame:
    """Q1-only summary (no Q2, SEM or NSP aggregations)."""

    summary_FSW = (
        ConsolidatedFSW
        .groupby(['county', 'sr_name', 'region'])
        .agg(
            TotalCohort=('kp_vp_unique_identifier_code', 'nunique'),
            TotalPE=('peer_educator_code', 'nunique'),

            HTSQ1=('hts_q1', 'sum'),
            HTS_Negative=('hts_Negative_q1', 'sum'),
            HTS_Positive=('hts_Positive_q1', 'sum'),

            DefinedPackage_Q1=('definedpackage_q1', 'sum'),

            PREP_Q1=('prep_q1', 'sum'),

            New_on_ART=('Positive_art', 'sum'),
            CurrentArt_Q1=('onart_q1', 'sum'),

            STISCREENEDQ_1=('sti_screened_q1', 'sum'),
            STIDiagnosid_Q1=('sti_diagnosed_q1', 'sum'),
            STItreated_Q1=('sti_treated_q1', 'sum'),
            TB_SCREENED_Q1=('tb_screened_q1', 'sum'),
            PeerEducation_Q1=('peereducation_q1', 'sum'),

            ReceivedCondom_Q1=('condoms_q1', 'sum'),
            total_condoms_Q1=('total_condoms_q1', 'sum'),
        )
        .reset_index()
    )

    return summary_FSW


def build_final_summary(
    ConsolidatedFSW: pd.DataFrame,
    summary_KPLHIV: pd.DataFrame
) -> pd.DataFrame:
    """
    Builds final_summary_FSW (Q1 only, no NSP) with your rules + forces template order.
    """

    # 1) Age group wide table
    age_group_counts = build_age_group_counts(ConsolidatedFSW)

    # 2) Base summary
    summary_FSW = build_summary_fsw(ConsolidatedFSW)
    final_df = summary_FSW.copy()

    # 3) Merge KPLHIV after base
    final_df = pd.merge(
        final_df,
        summary_KPLHIV,
        on=["county", "sr_name", "region"],
        how="left"
    )

    # 4) Merge age groups
    final_df = pd.merge(
        final_df,
        age_group_counts,
        on=["county", "sr_name", "region"],
        how="left"
    )

    # --- FORCE AGE GROUP ORDER LIKE TEMPLATE ---
    keys = ["county", "sr_name", "region"]
    age_order = ["15-19", "20-24", "25+", "Below 15"]

    # Ensure all age columns exist
    for c in age_order:
        if c not in final_df.columns:
            final_df[c] = 0

    # KPLHIV cols as in template
    kplhiv_cols = []
    for c in ["Total_KPLHIV", "Suppressed"]:
        if c in final_df.columns:
            kplhiv_cols.append(c)

    # --- TEMPLATE COLUMN ORDER (Q1 ONLY, NO NSP) ---
    template_order = (
        keys
        + age_order
        + ["TotalCohort", "TotalPE"]
        + ["HTSQ1", "HTS_Negative", "HTS_Positive"]
        + kplhiv_cols
        + [
            "DefinedPackage_Q1",
            "PREP_Q1",
            "New_on_ART", "CurrentArt_Q1",
            "STISCREENEDQ_1",
            "STIDiagnosid_Q1",
            "STItreated_Q1",
            "TB_SCREENED_Q1",
            "PeerEducation_Q1",
            "ReceivedCondom_Q1", "total_condoms_Q1",
        ]
    )

    # Keep only columns that exist, preserve order, avoid duplicates
    seen = set()
    ordered_cols = []
    for c in template_order:
        if c in final_df.columns and c not in seen:
            ordered_cols.append(c)
            seen.add(c)

    # Append any extra columns at the end (in case you have new indicators later)
    extras = [c for c in final_df.columns if c not in seen]
    final_df = final_df[ordered_cols + extras]

    # Uppercase demographics like you did
    demo_cols = ["county", "sr_name", "region", "key_vulnerable_population_type_"]
    existing_demo = [c for c in demo_cols if c in final_df.columns]
    for c in existing_demo:
        final_df[c] = final_df[c].astype(str).str.upper()

    return final_df


def export_like_template(final_df: pd.DataFrame, output_path: str):
    """
    Writes an Excel with 2-row merged headers exactly like the uploaded template (Q1 only, no NSP).
    Headers: Cascadia Code (bold), per-block colors from the template.
    Body/data + TOTAL row: Candara. Gridlines off.
    """

    # -------------------------
    # Header specification (Q1 only, no NSP)
    # -------------------------
    blocks = [
        ("COUNTY", ["COUNTY"], ["county"]),
        ("SR NAME", ["SR NAME"], ["sr_name"]),
        ("REGION", ["REGION"], ["region"]),

        ("AGE GROUP", ["15-19", "20-24", "25+", "Below 15"], ["15-19", "20-24", "25+", "Below 15"]),
        ("TOTAL", ["COHORT", "PE"], ["TotalCohort", "TotalPE"]),

        ("HIV TESTING & PLHIV",
         ["Q1", "NEGQ1", "POSQ1", "Total_KPLHIV", "Suppressed"],
         ["HTSQ1", "HTS_Negative", "HTS_Positive", "Total_KPLHIV", "Suppressed"]),

        ("DEFINED PACKAGE", ["Q1"], ["DefinedPackage_Q1"]),
        ("PREP ", ["Q1"], ["PREP_Q1"]),

        ("ART", ["NEWQ1", "CURRENTQ1"], ["New_on_ART", "CurrentArt_Q1"]),

        ("STI SCREENING", ["Q1"], ["STISCREENEDQ_1"]),
        ("STI DIAGNOSIS", ["Q1"], ["STIDiagnosid_Q1"]),
        ("STI TREATMENT", ["Q1"], ["STItreated_Q1"]),
        ("TB SCREENING", ["Q1"], ["TB_SCREENED_Q1"]),
        ("PEER EDUCATION", ["Q1"], ["PeerEducation_Q1"]),

        ("CONDOMS", ["Q1", "TOTQ1"], ["ReceivedCondom_Q1", "total_condoms_Q1"]),
    ]

    # Filter blocks to only columns present
    filtered_blocks = []
    for grp, subs, cols in blocks:
        keep = [(s, c) for s, c in zip(subs, cols) if c in final_df.columns]
        if keep:
            filtered_blocks.append((grp, [k[0] for k in keep], [k[1] for k in keep]))

    # -------------------------
    # Excel styling
    # -------------------------
    wb = Workbook()
    ws = wb.active
    ws.title = "KP_Summary"

    center = Alignment(horizontal="center", vertical="center", wrap_text=True)
    left = Alignment(horizontal="left", vertical="center", wrap_text=True)
    thin = Side(style="thin", color="000000")
    border = Border(left=thin, right=thin, top=thin, bottom=thin)

    # -------------------------
    # Write headers (2 rows)
    # -------------------------
    col_idx = 1
    col_map = []  # list of df columns in the final written order

    for grp, subs, cols in filtered_blocks:
        group_fill = PatternFill(
            "solid", fgColor=GROUP_HEADER_COLORS.get(grp, DEFAULT_GROUP_COLOR)
        )

        start = col_idx
        for sub, c in zip(subs, cols):
            # Row 2 subheader
            ws.cell(row=2, column=col_idx, value=sub)
            ws.cell(row=2, column=col_idx).font = HEADER_FONT
            ws.cell(row=2, column=col_idx).alignment = center
            ws.cell(row=2, column=col_idx).fill = SUBHEADER_FILL
            ws.cell(row=2, column=col_idx).border = border

            col_map.append(c)
            col_idx += 1

        end = col_idx - 1

        # Row 1 group header (merge across the block if block >1)
        if start == end:
            ws.cell(row=1, column=start, value=grp)
            ws.cell(row=1, column=start).font = HEADER_FONT
            ws.cell(row=1, column=start).alignment = center
            ws.cell(row=1, column=start).fill = group_fill
            ws.cell(row=1, column=start).border = border

            # Merge vertically for single-column blocks
            ws.merge_cells(start_row=1, start_column=start, end_row=2, end_column=end)
        else:
            ws.cell(row=1, column=start, value=grp)
            ws.cell(row=1, column=start).font = HEADER_FONT
            ws.cell(row=1, column=start).alignment = center
            ws.cell(row=1, column=start).fill = group_fill

            ws.merge_cells(start_row=1, start_column=start, end_row=1, end_column=end)

            # Apply header styling to merged region cells
            for c in range(start, end + 1):
                ws.cell(row=1, column=c).fill = group_fill
                ws.cell(row=1, column=c).border = border
                ws.cell(row=1, column=c).alignment = center
                ws.cell(row=1, column=c).font = HEADER_FONT

    # Freeze panes under headers
    ws.freeze_panes = "A3"

    # Kill the gridlines
    ws.sheet_view.showGridLines = False

    # -------------------------
    # Write data rows
    # -------------------------
    start_data_row = 3
    for i, row in enumerate(final_df[col_map].itertuples(index=False), start=start_data_row):
        for j, val in enumerate(row, start=1):
            cell = ws.cell(row=i, column=j, value=val)
            cell.border = border
            cell.alignment = left if j <= 3 else center
            cell.font = BODY_FONT

    last_data_row = start_data_row + len(final_df) - 1
    total_row = last_data_row + 1  # place totals immediately after the data

    # -------------------------
    # TOTAL row with formulas
    # -------------------------
    ws.merge_cells(start_row=total_row, start_column=1, end_row=total_row, end_column=3)
    tcell = ws.cell(row=total_row, column=1, value="TOTAL")
    tcell.font = BODY_FONT_BOLD
    tcell.alignment = center
    tcell.fill = TOTAL_FILL
    tcell.border = border

    for c in range(4, len(col_map) + 1):
        col_letter = get_column_letter(c)
        formula = f"=SUM({col_letter}{start_data_row}:{col_letter}{last_data_row})"
        cell = ws.cell(row=total_row, column=c, value=formula)
        cell.font = BODY_FONT_BOLD
        cell.alignment = center
        cell.fill = TOTAL_FILL
        cell.border = border

    # -------------------------
    # Column widths
    # -------------------------
    for c in range(1, len(col_map) + 1):
        ws.column_dimensions[get_column_letter(c)].width = 14 if c > 3 else 22

    ws.row_dimensions[1].height = 22
    ws.row_dimensions[2].height = 30

    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    wb.save(output_path)
    print(f"✅ Template-like summary exported: {output_path}")


# -------------------------
# USAGE EXAMPLE
# -------------------------
final_df = build_final_summary(ConsolidatedFSW, summary_KPLHIV)
export_like_template(final_df, "Summaries/KP Summary.xlsx")
final_df
```

```python
import os
import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import Alignment, Font, PatternFill, Border, Side
from openpyxl.utils import get_column_letter

# -------------------------
# Color constants (pulled from your original hardcoded hex — now you only
# change them in ONE place instead of grepping through the file like a caveman)
# -------------------------
SUBHEADER_FILL_COLOR = "FF7D81"   # coral — row 2, uniform across every block
TOTAL_FILL_COLOR = "6BD669"       # green — TOTAL row

# Row-1 group header colors — extracted directly from the uploaded template
# (some were Excel THEME colors with tint applied; already resolved to final RGB below)
GROUP_HEADER_COLORS = {
    "COUNTY": "00B050",
    "SR NAME": "00B050",
    "REGION": "00B050",
    "AGE GROUP": "B7DEE8",
    "TOTAL": "00B050",
    "HIV TESTING & PLHIV": "F79646",
    "DEFINED PACKAGE": "0070C0",
    "PREP ": "FCD5B5",
    "ART": "8064A2",
    "STI SCREENING": "EBF1DE",
    "STI DIAGNOSIS": "4BACC6",
    "STI TREATMENT": "FFC000",
    "TB SCREENING": "B3A2C7",
    "PEER EDUCATION": "FFFF00",
    "CONDOMS": "D9D9D9",
}
DEFAULT_GROUP_COLOR = "FF7D81"  # fallback if a block name isn't in the map above

SUBHEADER_FILL = PatternFill("solid", fgColor=SUBHEADER_FILL_COLOR)
TOTAL_FILL = PatternFill("solid", fgColor=TOTAL_FILL_COLOR)

# -------------------------
# Font constants
# -------------------------
HEADER_FONT_NAME = "Cascadia Code"
BODY_FONT_NAME = "Candara"

HEADER_FONT = Font(name=HEADER_FONT_NAME, bold=True)
BODY_FONT = Font(name=BODY_FONT_NAME)
BODY_FONT_BOLD = Font(name=BODY_FONT_NAME, bold=True)


def build_age_group_counts(df: pd.DataFrame) -> pd.DataFrame:
    """
    Returns wide age-group table:
    index = county, sr_name, region
    columns = age groups
    """
    age_group_counts = (
        df.groupby(["county", "sr_name", "region", "age_group"])
          .size()
          .reset_index(name="total")
          .pivot_table(
              index=["county", "sr_name", "region"],
              columns="age_group",
              values="total",
              fill_value=0,
              aggfunc="sum"
          )
          .reset_index()
    )
    return age_group_counts


def build_summary_fsw(ConsolidatedFSW: pd.DataFrame) -> pd.DataFrame:
    """
    Your main aggregation (FSW).
    NOTE: Assumes all the referenced columns already exist in ConsolidatedFSW.
    """
    summary_FSW = (
        ConsolidatedFSW
        .groupby(['county', 'sr_name', 'region'])
        .agg(
            TotalCohort=('kp_vp_unique_identifier_code', 'nunique'),
            TotalPE=('peer_educator_code', 'nunique'),
            HTSQ1=('hts_q1', 'sum'),
            HTSQ2=('hts_q2', 'sum'),
            HTS_SEM=('hts_sem', 'sum'),
            HTS_SEM_ANY=('hts_sem_any', 'sum'),
            HTS_Negative=('hts_Negative_q1', 'sum'),
            HTS_Negative_Q2=('hts_Negative_q2', 'sum'),
            HTS_Positive=('hts_Positive_q1', 'sum'),
            HTS_Positive_Q2=('hts_Positive_q2', 'sum'),

            DefinedPackage_Q1=('definedpackage_q1', 'sum'),
            DefinedPackage_Q2=('definedpackage_q2', 'sum'),
            DefinedPackage_SEM=('definedpackage_sem', 'sum'),

            PREP_Q1=('prep_q1', 'sum'),
            PREP_Q2=('prep_q2', 'sum'),
            PREP_SEM=('prep_sem', 'sum'),

            New_on_ART=('Positive_art', 'sum'),
            New_on_ART_Q2=('Positive_art_q2', 'sum'),
            CurrentArt_Q1=('onart_q1', 'sum'),
            CurrentArt_Q2=('onart_q2', 'sum'),

            STISCREENEDQ_1=('sti_screened_q1', 'sum'),
            STISCREENEDQ_2=('sti_screened_q2', 'sum'),
            STI_SCREENED_SEM=('sti_screened_sem', 'sum'),

            STIDiagnosid_Q1=('sti_diagnosed_q1', 'sum'),
            STI_DIAGNOSED_Q2=('sti_diagnosed_q2', 'sum'),
            STI_DIAGNOSED_SEM=('sti_diagnosed_sem', 'sum'),

            STItreated_Q1=('sti_treated_q1', 'sum'),
            STI_TREATED_Q2=('sti_treated_q2', 'sum'),
            STI_TREATED_SEM=('sti_treated_sem', 'sum'),

            TB_SCREENED_Q1=('tb_screened_q1', 'sum'),
            TB_SCREENED_Q2=('tb_screened_q2', 'sum'),
            TB_SCREENED_SEM=('tb_screened_sem', 'sum'),

            PeerEducation_Q1=('peereducation_q1', 'sum'),
            PeerEducation_Q2=('peereducation_q2', 'sum'),
            PeerEducation_SEM=('peereducation_sem', 'sum'),

            ReceivedCondom_Q1=('condoms_q1', 'sum'),
            ReceivedCondom_Q2=('condoms_q2', 'sum'),
            ReceivedCondom_SEM=('condoms_sem', 'sum'),

            total_condoms_Q1=('total_condoms_q1', 'sum'),
            total_condoms_Q2=('total_condoms_q2', 'sum'),
            total_condoms_SEM=('total_condoms_sem', 'sum')
        )
        .reset_index()
    )

    if "total_condoms_SEM" not in summary_FSW.columns:
        summary_FSW["total_condoms_SEM"] = (
            summary_FSW.get("total_condoms_Q1", 0) + summary_FSW.get("total_condoms_Q2", 0)
        )

    return summary_FSW


def build_final_summary(
    ConsolidatedFSW: pd.DataFrame,
    summary_KPLHIV: pd.DataFrame
) -> pd.DataFrame:
    """
    Builds final_summary_FSW with your rules + forces template order.
    """

    age_group_counts = build_age_group_counts(ConsolidatedFSW)
    summary_FSW = build_summary_fsw(ConsolidatedFSW)
    final_df = summary_FSW.copy()

    final_df = pd.merge(
        final_df,
        summary_KPLHIV,
        on=["county", "sr_name", "region"],
        how="left"
    )

    final_df = pd.merge(
        final_df,
        age_group_counts,
        on=["county", "sr_name", "region"],
        how="left"
    )

    keys = ["county", "sr_name", "region"]
    age_order = ["15-19", "20-24", "25+", "Below 15"]

    for c in age_order:
        if c not in final_df.columns:
            final_df[c] = 0

    kplhiv_cols = []
    for c in ["Total_KPLHIV", "Suppressed"]:
        if c in final_df.columns:
            kplhiv_cols.append(c)

    template_order = (
        keys
        + age_order
        + ["TotalCohort", "TotalPE"]
        + [
            "HTSQ1", "HTSQ2", "HTS_SEM", "HTS_SEM_ANY",
            "HTS_Negative", "HTS_Negative_Q2",
            "HTS_Positive", "HTS_Positive_Q2",
        ]
        + kplhiv_cols
        + [
            "DefinedPackage_Q1", "DefinedPackage_Q2", "DefinedPackage_SEM",
            "PREP_Q1", "PREP_Q2", "PREP_SEM",
            "New_on_ART", "New_on_ART_Q2", "CurrentArt_Q1", "CurrentArt_Q2",
            "STISCREENEDQ_1", "STISCREENEDQ_2", "STI_SCREENED_SEM",
            "STIDiagnosid_Q1", "STI_DIAGNOSED_Q2", "STI_DIAGNOSED_SEM",
            "STItreated_Q1", "STI_TREATED_Q2", "STI_TREATED_SEM",
            "TB_SCREENED_Q1", "TB_SCREENED_Q2", "TB_SCREENED_SEM",
            "PeerEducation_Q1", "PeerEducation_Q2", "PeerEducation_SEM",
            "ReceivedCondom_Q1", "ReceivedCondom_Q2", "ReceivedCondom_SEM",
            "total_condoms_Q1", "total_condoms_Q2", "total_condoms_SEM",
        ]
    )

    seen = set()
    ordered_cols = []
    for c in template_order:
        if c in final_df.columns and c not in seen:
            ordered_cols.append(c)
            seen.add(c)

    extras = [c for c in final_df.columns if c not in seen]
    final_df = final_df[ordered_cols + extras]

    demo_cols = ["county", "sr_name", "region", "key_vulnerable_population_type_"]
    existing_demo = [c for c in demo_cols if c in final_df.columns]
    for c in existing_demo:
        final_df[c] = final_df[c].astype(str).str.upper()

    return final_df


def export_like_template(final_df: pd.DataFrame, output_path: str):
    """
    Writes an Excel with 2-row merged headers exactly like the uploaded template.
    Headers: Cascadia Code (bold). Body/data + TOTAL row: Candara.
    """

    blocks = [
        ("COUNTY", ["COUNTY"], ["county"]),
        ("SR NAME", ["SR NAME"], ["sr_name"]),
        ("REGION", ["REGION"], ["region"]),

        ("AGE GROUP", ["15-19", "20-24", "25+", "Below 15"], ["15-19", "20-24", "25+", "Below 15", "TotalCohort"]),
        ("TOTAL", ["COHORT", "PE"], ["TotalCohort", "TotalPE"]),

        ("HIV TESTING & PLHIV",
         ["Q1", "Q2", "SEM", "SEM_ANY", "NEGQ1", "NEGQ2", "POSQ1", "POSQ2", "Total_KPLHIV", "Suppressed"],
         ["HTSQ1", "HTSQ2", "HTS_SEM", "HTS_SEM_ANY", "HTS_Negative", "HTS_Negative_Q2", "HTS_Positive",
          "HTS_Positive_Q2", "Total_KPLHIV", "Suppressed"]
        ),

        ("DEFINED PACKAGE", ["Q1", "Q2", "SEM"], ["DefinedPackage_Q1", "DefinedPackage_Q2", "DefinedPackage_SEM"]),
        ("PREP ", ["Q1", "Q2", "SEM"], ["PREP_Q1", "PREP_Q2", "PREP_SEM"]),

        ("ART", ["NEWQ1", "NEWQ2", "CURRENTQ1", "CURRENTQ2"],
         ["New_on_ART", "New_on_ART_Q2", "CurrentArt_Q1", "CurrentArt_Q2"]),

        ("STI SCREENING", ["Q1", "Q2", "SEM"], ["STISCREENEDQ_1", "STISCREENEDQ_2", "STI_SCREENED_SEM"]),
        ("STI DIAGNOSIS", ["Q1", "Q2", "SEM"], ["STIDiagnosid_Q1", "STI_DIAGNOSED_Q2", "STI_DIAGNOSED_SEM"]),
        ("STI TREATMENT", ["Q1", "Q2", "SEM"], ["STItreated_Q1", "STI_TREATED_Q2", "STI_TREATED_SEM"]),
        ("TB SCREENING", ["Q1", "Q2", "SEM"], ["TB_SCREENED_Q1", "TB_SCREENED_Q2", "TB_SCREENED_SEM"]),
        ("PEER EDUCATION", ["Q1", "Q2", "SEM"], ["PeerEducation_Q1", "PeerEducation_Q2", "PeerEducation_SEM"]),

        ("CONDOMS", ["Q1", "Q2", "SEM", "TOTQ1", "TOTQ2", "SEM"],
         ["ReceivedCondom_Q1", "ReceivedCondom_Q2", "ReceivedCondom_SEM",
          "total_condoms_Q1", "total_condoms_Q2", "total_condoms_SEM"]),
    ]

    filtered_blocks = []
    for grp, subs, cols in blocks:
        keep = [(s, c) for s, c in zip(subs, cols) if c in final_df.columns]
        if keep:
            filtered_blocks.append((grp, [k[0] for k in keep], [k[1] for k in keep]))

    wb = Workbook()
    ws = wb.active
    ws.title = "Summary"

    center = Alignment(horizontal="center", vertical="center", wrap_text=True)
    left = Alignment(horizontal="left", vertical="center", wrap_text=True)
    thin = Side(style="thin", color="000000")
    border = Border(left=thin, right=thin, top=thin, bottom=thin)

    # -------------------------
    # Write headers (2 rows) — Cascadia Code, bold, coral fill
    # -------------------------
    col_idx = 1
    col_map = []

    for grp, subs, cols in filtered_blocks:
        group_fill = PatternFill(
            "solid", fgColor=GROUP_HEADER_COLORS.get(grp, DEFAULT_GROUP_COLOR)
        )

        start = col_idx
        for sub, c in zip(subs, cols):
            cell = ws.cell(row=2, column=col_idx, value=sub)
            cell.font = HEADER_FONT
            cell.alignment = center
            cell.fill = SUBHEADER_FILL
            cell.border = border

            col_map.append(c)
            col_idx += 1

        end = col_idx - 1

        if start == end:
            cell = ws.cell(row=1, column=start, value=grp)
            cell.font = HEADER_FONT
            cell.alignment = center
            cell.fill = group_fill
            cell.border = border
            ws.merge_cells(start_row=1, start_column=start, end_row=2, end_column=end)
        else:
            cell = ws.cell(row=1, column=start, value=grp)
            cell.font = HEADER_FONT
            cell.alignment = center
            cell.fill = group_fill
            ws.merge_cells(start_row=1, start_column=start, end_row=1, end_column=end)

            for c in range(start, end + 1):
                ws.cell(row=1, column=c).fill = group_fill
                ws.cell(row=1, column=c).border = border
                ws.cell(row=1, column=c).alignment = center
                ws.cell(row=1, column=c).font = HEADER_FONT

    ws.freeze_panes = "A3"
    ws.sheet_view.showGridLines = False
    # -------------------------
    # Write data rows — Candara, regular weight
    # -------------------------
    start_data_row = 3
    for i, row in enumerate(final_df[col_map].itertuples(index=False), start=start_data_row):
        for j, val in enumerate(row, start=1):
            cell = ws.cell(row=i, column=j, value=val)
            cell.border = border
            cell.alignment = left if j <= 3 else center
            cell.font = BODY_FONT

    last_data_row = start_data_row + len(final_df) - 1
    total_row = last_data_row + 1

    # -------------------------
    # TOTAL row — Candara, bold, green fill
    # -------------------------
    ws.merge_cells(start_row=total_row, start_column=1, end_row=total_row, end_column=3)
    tcell = ws.cell(row=total_row, column=1, value="TOTAL")
    tcell.font = BODY_FONT_BOLD
    tcell.alignment = center
    tcell.fill = TOTAL_FILL
    tcell.border = border

    for c in range(4, len(col_map) + 1):
        col_letter = get_column_letter(c)
        formula = f"=SUM({col_letter}{start_data_row}:{col_letter}{last_data_row})"
        cell = ws.cell(row=total_row, column=c, value=formula)
        cell.font = BODY_FONT_BOLD
        cell.alignment = center
        cell.fill = TOTAL_FILL
        cell.border = border

    for c in range(1, len(col_map) + 1):
        ws.column_dimensions[get_column_letter(c)].width = 14 if c > 3 else 22

    ws.row_dimensions[1].height = 22
    ws.row_dimensions[2].height = 30

    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    wb.save(output_path)
    print(f"✅ Template-like summary exported: {output_path}")


# -------------------------
# USAGE EXAMPLE
# -------------------------
final_df = build_final_summary(ConsolidatedFSW, summary_KPLHIV)
export_like_template(final_df, "Summaries/summary KP.xlsx")
display(final_df)
```

```python
import os
import pandas as pd
from openpyxl import Workbook
from openpyxl.styles import Alignment, Font, PatternFill, Border, Side
from openpyxl.utils import get_column_letter

# -------------------------
# Color constants — resolved from the actual FSW template (theme colors + tint decoded)
# -------------------------
SUBHEADER_FILL_COLOR = "FF7D81"   # coral — row 2, uniform across every block
TOTAL_FILL_COLOR = "6BD669"       # green — TOTAL row

GROUP_HEADER_COLORS = {
    "COUNTY": "00B050",
    "SR NAME": "00B050",
    "REGION": "00B050",
    "AGE GROUP": "B7DEE8",
    "TOTAL": "00B050",
    "HIV TESTING & PLHIV": "F79646",
    "DEFINED PACKAGE": "0070C0",
    "PREP ": "FCD5B5",
    "ART": "8064A2",
    "STI SCREENING": "EBF1DE",
    "STI DIAGNOSIS": "4BACC6",
    "STI TREATMENT": "FFC000",
    "TB SCREENING": "B3A2C7",
    "PEER EDUCATION": "FFFF00",
    "CONDOMS": "D9D9D9",
    # NSP wasn't in the original template — no real reference color exists yet.
    # Falling back to subheader coral until someone hands you an actual spec.
    "NSP": "FF7D81",
}
DEFAULT_GROUP_COLOR = "FF7D81"  # fallback for any block name not in the map above

SUBHEADER_FILL = PatternFill("solid", fgColor=SUBHEADER_FILL_COLOR)
TOTAL_FILL = PatternFill("solid", fgColor=TOTAL_FILL_COLOR)

# -------------------------
# Font constants
# -------------------------
HEADER_FONT_NAME = "Cascadia Code"
BODY_FONT_NAME = "Candara"

HEADER_FONT = Font(name=HEADER_FONT_NAME, bold=True)
BODY_FONT = Font(name=BODY_FONT_NAME)
BODY_FONT_BOLD = Font(name=BODY_FONT_NAME, bold=True)


def build_age_group_counts(df: pd.DataFrame) -> pd.DataFrame:

    age_group_counts = (
        df.groupby(["county", "sr_name", "region", "age_group"])
          .size()
          .reset_index(name="total")
          .pivot_table(
              index=["county", "sr_name", "region"],
              columns="age_group",
              values="total",
              fill_value=0,
              aggfunc="sum"
          )
          .reset_index()
    )
    return age_group_counts


def build_summary_fsw(ConsolidatedFSW: pd.DataFrame) -> pd.DataFrame:

    summary_FSW = (
        ConsolidatedFSW
        .groupby(['county', 'sr_name', 'region'])
        .agg(
            TotalCohort=('kp_vp_unique_identifier_code', 'nunique'),
            TotalPE=('peer_educator_code', 'nunique'),
            HTSQ1=('hts_q1', 'sum'),
            HTSQ2=('hts_q2', 'sum'),
            HTS_SEM=('hts_sem', 'sum'),
            HTS_SEM_ANY=('hts_sem_any', 'sum'),
            HTS_Negative=('hts_Negative_q1', 'sum'),
            HTS_Negative_Q2=('hts_Negative_q2', 'sum'),
            HTS_Positive=('hts_Positive_q1', 'sum'),
            HTS_Positive_Q2=('hts_Positive_q2', 'sum'),

            DefinedPackage_Q1=('definedpackage_q1', 'sum'),
            DefinedPackage_Q2=('definedpackage_q2', 'sum'),
            DefinedPackage_SEM=('definedpackage_sem', 'sum'),

            PREP_Q1=('prep_q1', 'sum'),
            PREP_Q2=('prep_q2', 'sum'),
            PREP_SEM=('prep_sem', 'sum'),

            New_on_ART=('Positive_art', 'sum'),
            New_on_ART_Q2=('Positive_art_q2', 'sum'),
            CurrentArt_Q1=('onart_q1', 'sum'),
            CurrentArt_Q2=('onart_q2', 'sum'),

            STISCREENEDQ_1=('sti_screened_q1', 'sum'),
            STISCREENEDQ_2=('sti_screened_q2', 'sum'),
            STI_SCREENED_SEM=('sti_screened_sem', 'sum'),

            STIDiagnosid_Q1=('sti_diagnosed_q1', 'sum'),
            STI_DIAGNOSED_Q2=('sti_diagnosed_q2', 'sum'),
            STI_DIAGNOSED_SEM=('sti_diagnosed_sem', 'sum'),

            STItreated_Q1=('sti_treated_q1', 'sum'),
            STI_TREATED_Q2=('sti_treated_q2', 'sum'),
            STI_TREATED_SEM=('sti_treated_sem', 'sum'),

            TB_SCREENED_Q1=('tb_screened_q1', 'sum'),
            TB_SCREENED_Q2=('tb_screened_q2', 'sum'),
            TB_SCREENED_SEM=('tb_screened_sem', 'sum'),

            PeerEducation_Q1=('peereducation_q1', 'sum'),
            PeerEducation_Q2=('peereducation_q2', 'sum'),
            PeerEducation_SEM=('peereducation_sem', 'sum'),

            ReceivedCondom_Q1=('condoms_q1', 'sum'),
            ReceivedCondom_Q2=('condoms_q2', 'sum'),
            ReceivedCondom_SEM=('condoms_sem', 'sum'),

            NSP_Q1=('nsp_q1', 'sum'),
            NSP_Q2=('nsp_q2', 'sum'),
            NSP_SEM=('nsp_sem', 'sum'),

            total_condoms_Q1=('total_condoms_q1', 'sum'),
            total_condoms_Q2=('total_condoms_q2', 'sum'),
            total_condoms_SEM=('total_condoms_sem', 'sum'),
            total_nsp_Q1=('total_nsp_q1', 'sum'),
            total_nsp_Q2=('total_nsp_q2', 'sum')
        )
        .reset_index()
    )

    # If total_condoms_sem doesn't exist upstream, compute it safely:
    if "total_condoms_SEM" not in summary_FSW.columns:
        summary_FSW["total_condoms_SEM"] = (
            summary_FSW.get("total_condoms_Q1", 0) + summary_FSW.get("total_condoms_Q2", 0)
        )

    return summary_FSW


def build_final_summary(
    ConsolidatedFSW: pd.DataFrame,
    # cohort_summary: pd.DataFrame,
    summary_KPLHIV: pd.DataFrame
) -> pd.DataFrame:
    """
    Builds final_summary_FSW with your rules + forces template order.
    """

    # 1) Age group wide table
    age_group_counts = build_age_group_counts(ConsolidatedFSW)

    # 2) Base summary
    summary_FSW = build_summary_fsw(ConsolidatedFSW)
    final_df = summary_FSW.copy()

    # 4) Merge KPLHIV after base
    final_df = pd.merge(
        final_df,
        summary_KPLHIV,
        on=["county", "sr_name", "region"],
        how="left"
    )

    # 5) Merge age groups
    final_df = pd.merge(
        final_df,
        age_group_counts,
        on=["county", "sr_name", "region"],
        how="left"
    )

    # --- FORCE AGE GROUP ORDER LIKE TEMPLATE ---
    keys = ["county", "sr_name", "region"]
    age_order = ["15-19", "20-24", "25+", "Below 15"]

    # Ensure all age columns exist
    for c in age_order:
        if c not in final_df.columns:
            final_df[c] = 0

    # KPLHIV cols as in template
    kplhiv_cols = []
    for c in ["Total_KPLHIV", "Suppressed"]:
        if c in final_df.columns:
            kplhiv_cols.append(c)

    # --- TEMPLATE COLUMN ORDER (EXACT STRUCTURE) ---
    template_order = (
        keys
        + age_order
        + ["TotalCohort", "TotalPE"]
        + [
            "HTSQ1", "HTSQ2", "HTS_SEM", "HTS_SEM_ANY",
            "HTS_Negative", "HTS_Negative_Q2",
            "HTS_Positive", "HTS_Positive_Q2",
        ]
        + kplhiv_cols
        + [
            "DefinedPackage_Q1", "DefinedPackage_Q2", "DefinedPackage_SEM",
            "PREP_Q1", "PREP_Q2", "PREP_SEM",
            "New_on_ART", "New_on_ART_Q2", "CurrentArt_Q1", "CurrentArt_Q2",
            "STISCREENEDQ_1", "STISCREENEDQ_2", "STI_SCREENED_SEM",
            "STIDiagnosid_Q1", "STI_DIAGNOSED_Q2", "STI_DIAGNOSED_SEM",
            "STItreated_Q1", "STI_TREATED_Q2", "STI_TREATED_SEM",
            "TB_SCREENED_Q1", "TB_SCREENED_Q2", "TB_SCREENED_SEM",
            "PeerEducation_Q1", "PeerEducation_Q2", "PeerEducation_SEM",
            "ReceivedCondom_Q1", "ReceivedCondom_Q2", "ReceivedCondom_SEM",
            "total_condoms_Q1", "total_condoms_Q2", "total_condoms_SEM",
             # ✅ NSP block
            "NSP_Q1", "NSP_Q2", "NSP_SEM",
            "total_nsp_Q1", "total_nsp_Q2", "total_nsp_SEM",
        ]
    )

    # Keep only columns that exist, preserve order, avoid duplicates
    seen = set()
    ordered_cols = []
    for c in template_order:
        if c in final_df.columns and c not in seen:
            ordered_cols.append(c)
            seen.add(c)

    # Append any extra columns at the end (in case you have new indicators later)
    extras = [c for c in final_df.columns if c not in seen]
    final_df = final_df[ordered_cols + extras]

    # Uppercase demographics like you did
    demo_cols = ["county", "sr_name", "region", "key_vulnerable_population_type_"]
    existing_demo = [c for c in demo_cols if c in final_df.columns]
    for c in existing_demo:
        final_df[c] = final_df[c].astype(str).str.upper()

    return final_df


def export_like_template(final_df: pd.DataFrame, output_path: str):
    """
    Writes an Excel with 2-row merged headers exactly like the uploaded template.
    Headers: Cascadia Code (bold), per-block colors from the template.
    Body/data + TOTAL row: Candara. Gridlines off.
    """

    # -------------------------
    # Header specification
    # -------------------------
    # Column blocks: (Group Title, [subheaders], [data columns])
    blocks = [
        ("COUNTY", ["COUNTY"], ["county"]),
        ("SR NAME", ["SR NAME"], ["sr_name"]),
        ("REGION", ["REGION"], ["region"]),

        ("AGE GROUP", ["15-19", "20-24", "25+", "Below 15"], ["15-19", "20-24", "25+", "Below 15"]),
        ("TOTAL", ["COHORT", "PE"], ["TotalCohort", "TotalPE"]),

        ("HIV TESTING & PLHIV",
         ["Q1", "Q2", "SEM", "SEM_ANY", "NEGQ1", "NEGQ2", "POSQ1", "POSQ2", "Total_KPLHIV", "Suppressed"],
         ["HTSQ1", "HTSQ2", "HTS_SEM", "HTS_SEM_ANY", "HTS_Negative", "HTS_Negative_Q2", "HTS_Positive",
          "HTS_Positive_Q2", "Total_KPLHIV", "Suppressed"]
        ),

        ("DEFINED PACKAGE", ["Q1", "Q2", "SEM"], ["DefinedPackage_Q1", "DefinedPackage_Q2", "DefinedPackage_SEM"]),
        ("PREP ", ["Q1", "Q2", "SEM"], ["PREP_Q1", "PREP_Q2", "PREP_SEM"]),

        ("ART", ["NEWQ1", "NEWQ2", "CURRENTQ1", "CURRENTQ2"],
         ["New_on_ART", "New_on_ART_Q2", "CurrentArt_Q1", "CurrentArt_Q2"]),

        ("STI SCREENING", ["Q1", "Q2", "SEM"], ["STISCREENEDQ_1", "STISCREENEDQ_2", "STI_SCREENED_SEM"]),
        ("STI DIAGNOSIS", ["Q1", "Q2", "SEM"], ["STIDiagnosid_Q1", "STI_DIAGNOSED_Q2", "STI_DIAGNOSED_SEM"]),
        ("STI TREATMENT", ["Q1", "Q2", "SEM"], ["STItreated_Q1", "STI_TREATED_Q2", "STI_TREATED_SEM"]),
        ("TB SCREENING", ["Q1", "Q2", "SEM"], ["TB_SCREENED_Q1", "TB_SCREENED_Q2", "TB_SCREENED_SEM"]),
        ("PEER EDUCATION", ["Q1", "Q2", "SEM"], ["PeerEducation_Q1", "PeerEducation_Q2", "PeerEducation_SEM"]),

        ("CONDOMS", ["Q1", "Q2", "SEM", "TOTQ1", "TOTQ2", "SEM"],
         ["ReceivedCondom_Q1", "ReceivedCondom_Q2", "ReceivedCondom_SEM",
          "total_condoms_Q1", "total_condoms_Q2", "total_condoms_SEM"]),
        ("NSP", ["Q1", "Q2", "SEM", "TOTQ1", "TOTQ2", "SEM"],
        ["NSP_Q1", "NSP_Q2", "NSP_SEM",
         "total_nsp_Q1", "total_nsp_Q2", "total_nsp_SEM"]),]

    # Filter blocks to only columns present
    filtered_blocks = []
    for grp, subs, cols in blocks:
        keep = [(s, c) for s, c in zip(subs, cols) if c in final_df.columns]
        if keep:
            filtered_blocks.append((grp, [k[0] for k in keep], [k[1] for k in keep]))

    # -------------------------
    # Excel styling
    # -------------------------
    wb = Workbook()
    ws = wb.active
    ws.title = "PWID_Summary"

    center = Alignment(horizontal="center", vertical="center", wrap_text=True)
    left = Alignment(horizontal="left", vertical="center", wrap_text=True)
    thin = Side(style="thin", color="000000")
    border = Border(left=thin, right=thin, top=thin, bottom=thin)

    # -------------------------
    # Write headers (2 rows)
    # -------------------------
    col_idx = 1
    col_map = []  # list of df columns in the final written order

    for grp, subs, cols in filtered_blocks:
        group_fill = PatternFill(
            "solid", fgColor=GROUP_HEADER_COLORS.get(grp, DEFAULT_GROUP_COLOR)
        )

        start = col_idx
        for sub, c in zip(subs, cols):
            # Row 2 subheader
            ws.cell(row=2, column=col_idx, value=sub)
            ws.cell(row=2, column=col_idx).font = HEADER_FONT
            ws.cell(row=2, column=col_idx).alignment = center
            ws.cell(row=2, column=col_idx).fill = SUBHEADER_FILL
            ws.cell(row=2, column=col_idx).border = border

            col_map.append(c)
            col_idx += 1

        end = col_idx - 1

        # Row 1 group header (merge across the block if block >1)
        if start == end:
            ws.cell(row=1, column=start, value=grp)
            ws.cell(row=1, column=start).font = HEADER_FONT
            ws.cell(row=1, column=start).alignment = center
            ws.cell(row=1, column=start).fill = group_fill
            ws.cell(row=1, column=start).border = border

            # Merge vertically for single header columns like COUNTY/SR/REGION
            ws.merge_cells(start_row=1, start_column=start, end_row=2, end_column=end)
        else:
            ws.cell(row=1, column=start, value=grp)
            ws.cell(row=1, column=start).font = HEADER_FONT
            ws.cell(row=1, column=start).alignment = center
            ws.cell(row=1, column=start).fill = group_fill

            ws.merge_cells(start_row=1, start_column=start, end_row=1, end_column=end)

            # Apply header styling to merged region cells
            for c in range(start, end + 1):
                ws.cell(row=1, column=c).fill = group_fill
                ws.cell(row=1, column=c).border = border
                ws.cell(row=1, column=c).alignment = center
                ws.cell(row=1, column=c).font = HEADER_FONT

    # Freeze panes under headers
    ws.freeze_panes = "A3"

    # Kill the gridlines
    ws.sheet_view.showGridLines = False

    # -------------------------
    # Write data rows
    # -------------------------
    start_data_row = 3
    for i, row in enumerate(final_df[col_map].itertuples(index=False), start=start_data_row):
        for j, val in enumerate(row, start=1):
            cell = ws.cell(row=i, column=j, value=val)
            cell.border = border
            cell.alignment = left if j <= 3 else center
            cell.font = BODY_FONT

    last_data_row = start_data_row + len(final_df) - 1
    total_row = last_data_row + 1  # place totals immediately after the data
    # -------------------------
    # TOTAL row with formulas
    # -------------------------
    ws.merge_cells(start_row=total_row, start_column=1, end_row=total_row, end_column=3)
    tcell = ws.cell(row=total_row, column=1, value="TOTAL")
    tcell.font = BODY_FONT_BOLD
    tcell.alignment = center
    tcell.fill = TOTAL_FILL
    tcell.border = border

    for c in range(4, len(col_map) + 1):
        col_letter = get_column_letter(c)
        formula = f"=SUM({col_letter}{start_data_row}:{col_letter}{last_data_row})"
        cell = ws.cell(row=total_row, column=c, value=formula)
        cell.font = BODY_FONT_BOLD
        cell.alignment = center
        cell.fill = TOTAL_FILL
        cell.border = border

    # -------------------------
    # Column widths (auto-ish)
    # -------------------------
    for c in range(1, len(col_map) + 1):
        ws.column_dimensions[get_column_letter(c)].width = 14 if c > 3 else 22

    ws.row_dimensions[1].height = 22
    ws.row_dimensions[2].height = 30

    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    wb.save(output_path)
    print(f"✅ Template-like summary exported: {output_path}")


# -------------------------
# USAGE EXAMPLE
# -------------------------
final_df = build_final_summary(ConsolidatedFSW, summary_KPLHIV )
export_like_template(final_df, "Summaries/PWID Summary.xlsx")
final_df
```

```python
import pandas as pd
import numpy as np
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter
from pathlib import Path
from datetime import datetime
# Work on a copy so the original cohort data remains unchanged
kvp_merged = ConsolidatedFSW.copy()

# -------------------------------
# 1. Validate coverage indicators
# -------------------------------
coverage_columns = [
    'prep_q1',
    'prep_q2',
    'prep_sem',
    'hts_q1',
    'hts_q2',
    'hts_sem_any',
    'definedpackage_q1',
    'definedpackage_q2',
    'definedpackage_sem'
]

missing_columns = [
    column for column in coverage_columns
    if column not in kvp_merged.columns
]

if missing_columns:
    raise KeyError(
        f"Run the coverage indicator calculation first. "
        f"Missing columns: {missing_columns}"
    )

# -------------------------------
# 2. Ensure numeric age
# -------------------------------
kvp_merged['age'] = pd.to_numeric(
    kvp_merged['age'],
    errors='coerce'
)

# -------------------------------
# 3. Create age groups
# -------------------------------
def assign_age_group(age):
    if pd.isna(age):
        return 'Unknown'
    if 15 <= age <= 19:
        return '15-19'
    if 20 <= age <= 24:
        return '20-24'
    if age >= 25:
        return '25+'
    return 'Unknown'


kvp_merged['age_group'] = kvp_merged['age'].apply(assign_age_group)

# -------------------------------
# 4. Standardize sex values
# -------------------------------
kvp_merged['sex_group'] = (
    kvp_merged['sex_m_f_']
    .astype('string')
    .str.strip()
    .str.upper()
    .replace({
        'MALE': 'M',
        'FEMALE': 'F'
    })
    .fillna('F')
)

# -------------------------------
# 5. Create combined age-sex group
# -------------------------------
kvp_merged['age_sex_group'] = (
    kvp_merged['age_group']
    + '_'
    + kvp_merged['sex_group']
 )

# -------------------------------
# 6. Use coverage indicator columns
# -------------------------------
kvp_merged['PrEP_SEM'] = kvp_merged['prep_sem'].astype(int)
kvp_merged['HTS_Tested'] = kvp_merged['hts_sem_any'].astype(int)
kvp_merged['DefinedPackage_SEM'] = (
    kvp_merged['definedpackage_sem'].astype(int)
)

# -------------------------------
# 7. Define grouping columns
# -------------------------------
group_cols = [
    'sr_name',
    'region',
    'county',
    # 'key_vulnerable_population_type_'
]

for column in group_cols:
    kvp_merged[column] = (
        kvp_merged[column]
        .astype('string')
        .str.strip()
        .str.upper()
    )

missing_group_columns = [
    column for column in group_cols
    if column not in kvp_merged.columns
]

if missing_group_columns:
    raise KeyError(
        f"Missing grouping columns: {missing_group_columns}"
    )
# ---------------------------------------------------------
# 8. Indicator blocks and age-sex order
# ---------------------------------------------------------
age_sex_order = [
    '15-19_M', '15-19_F',
    '20-24_M', '20-24_F',
    '25+_M', '25+_F',
]

indicator_blocks = {
    'DefinedPackage_SEM': 'DEFINED PACKAGE',
    'HTS_Tested': 'HTS',
    'PrEP_SEM': 'PREP '
}

indicator_columns = list(indicator_blocks)

# ---------------------------------------------------------
# 9. Build one block per indicator
# ---------------------------------------------------------
block_tables = []

for indicator in indicator_columns:
    block = kvp_merged.pivot_table(
        index=group_cols,
        columns='age_sex_group',
        values=indicator,
        aggfunc='sum',
        fill_value=0
    )

    block = block.reindex(columns=age_sex_order, fill_value=0)
    block.columns = [
        f'{indicator}_{age_sex}'
        for age_sex in block.columns
    ]

    block[f'{indicator}_Total'] = block.sum(axis=1)
    block_tables.append(block)

summary = (
    pd.concat(block_tables, axis=1)
    .reset_index()
)

# Explicit column order
ordered_columns = group_cols.copy()

for indicator in indicator_columns:
    ordered_columns.extend([
        f'{indicator}_{age_sex}'
        for age_sex in age_sex_order
    ])
    ordered_columns.append(f'{indicator}_Total')

summary = summary[
    [column for column in ordered_columns if column in summary.columns]
]

pd.set_option('display.max_columns', None)
display(summary)

# ---------------------------------------------------------
# 10. Export with grouped indicator headers
# ---------------------------------------------------------
output_dir = Path(directory).resolve()
output_dir.mkdir(parents=True, exist_ok=True)

output_file = output_dir / 'KVP_Summary.xlsx'

if output_file.exists():
    try:
        with open(output_file, 'ab'):
            pass
    except OSError:
        timestamp = datetime.now().strftime('%Y%m%d_%H%M%S')
        output_file = output_dir / f'KVP_Summary_{timestamp}.xlsx'

thin = Side(style='thin', color='B0B0B0')
border = Border(left=thin, right=thin, top=thin, bottom=thin)
center = Alignment(
    horizontal='center',
    vertical='center',
    wrap_text=True
)

with pd.ExcelWriter(str(output_file), engine='openpyxl') as writer:
    summary.to_excel(
        writer,
        index=False,
        sheet_name='Summary',
        startrow=1
    )

    ws = writer.sheets['Summary']
    n_group_cols = len(group_cols)

    # Location/KP header block
    ws.merge_cells(
        start_row=1,
        start_column=1,
        end_row=1,
        end_column=n_group_cols
    )

    location_cell = ws.cell(
        row=1,
        column=1,
        value='DEMOGRAPHICS'
    )
    location_cell.font = Font(bold=True, color='FFFFFF')
    location_cell.fill = PatternFill(
        'solid',
        fgColor=GROUP_HEADER_COLORS['SR NAME']
    )
    location_cell.alignment = center
    location_cell.border = border

    # Style location columns
    for column_index in range(1, n_group_cols + 1):
        cell = ws.cell(row=2, column=column_index)
        cell.font = Font(bold=True)
        cell.alignment = center
        cell.border = border
        cell.fill = PatternFill(
            'solid',
            fgColor=GROUP_HEADER_COLORS['SR NAME']
        )

    # Indicator blocks
    column_cursor = n_group_cols + 1

    for indicator, block_name in indicator_blocks.items():
        block_columns = [
            f'{indicator}_{age_sex}'
            for age_sex in age_sex_order
        ]
        block_columns.append(f'{indicator}_Total')

        block_columns = [
            column for column in block_columns
            if column in summary.columns
        ]

        if not block_columns:
            continue

        start_column = column_cursor
        end_column = column_cursor + len(block_columns) - 1
        fill_color = GROUP_HEADER_COLORS[block_name]

        ws.merge_cells(
            start_row=1,
            start_column=start_column,
            end_row=1,
            end_column=end_column
        )

        block_cell = ws.cell(
            row=1,
            column=start_column,
            value=block_name.strip()
        )
        block_cell.font = Font(bold=True, color='FFFFFF')
        block_cell.fill = PatternFill(
            'solid',
            fgColor=fill_color
        )
        block_cell.alignment = center
        block_cell.border = border

        for column_index in range(start_column, end_column + 1):
            header_cell = ws.cell(row=2, column=column_index)
            header_cell.font = Font(bold=True)
            header_cell.alignment = center
            header_cell.border = border
            header_cell.fill = PatternFill(
                'solid',
                fgColor=fill_color
            )

        column_cursor = end_column + 1

    # Data rows
    for row_index in range(3, len(summary) + 3):
        for column_index in range(1, summary.shape[1] + 1):
            cell = ws.cell(row=row_index, column=column_index)
            cell.border = border

            if column_index > n_group_cols:
                cell.alignment = Alignment(horizontal='center')

            if row_index % 2 == 0:
                cell.fill = PatternFill(
                    'solid',
                    fgColor='F2F2F2'
                )

    # Bold totals
    for column_index, column_name in enumerate(summary.columns, start=1):
        if column_name.endswith('_Total'):
            for row_index in range(3, len(summary) + 3):
                ws.cell(
                    row=row_index,
                    column=column_index
                ).font = Font(bold=True)

    # Column widths
    for column_index, column_name in enumerate(summary.columns, start=1):
        max_length = max(
            len(str(column_name)),
            *(len(str(value)) for value in summary.iloc[:, column_index - 1])
        )
        ws.column_dimensions[
            get_column_letter(column_index)
        ].width = min(max_length + 3, 24)

    ws.freeze_panes = 'A3'
    ws.sheet_view.showGridLines = False
    ws.row_dimensions[1].height = 24
    ws.row_dimensions[2].height = 36

print(f'Summary saved successfully to: {output_file}')
```
